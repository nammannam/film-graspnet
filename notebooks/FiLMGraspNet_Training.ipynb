```python

from pathlib import Path
import shutil
import subprocess

DRIVE = Path("/content/drive/MyDrive/0.InterviewPrep/FPT-Innovators/Assignment/FiLM-GraspNet")
PROJECT = Path("/content/film-graspnet")
DATA = Path("/content/data")

DATA.mkdir(parents=True, exist_ok=True)

# Extract pilot datasets to Colab local storage
for name in ("train_pilot", "val_pilot"):
    archive = DRIVE / "transfer" / f"{name}.tar.gz"

    subprocess.run(
        ["tar", "-xzf", str(archive), "-C", str(DATA)],
        check=True,
    )

    print(f"{name}: EXTRACTED")

# Copy pretrained GR-ConvNet3 weights
checkpoint_dir = PROJECT / "checkpoints"
checkpoint_dir.mkdir(parents=True, exist_ok=True)

shutil.copy2(
    DRIVE / "transfer" / "grconvnet3_pretrained.pth",
    checkpoint_dir / "grconvnet3_pretrained.pth",
)

print("COLAB DATA & CHECKPOINT PREPARATION PASSED")
```

```python

import torch
from filmgrasp.data.dataset import GraspAnythingPPDataset

train_ds = GraspAnythingPPDataset("/content/data/train_pilot")
val_ds = GraspAnythingPPDataset("/content/data/val_pilot")

assert len(train_ds) == 381
assert len(val_ds) == 76

for name, dataset in [
    ("Train", train_ds),
    ("Validation", val_ds),
]:
    sample = dataset[0]

    assert sample["image"].shape == (3, 416, 416)
    assert torch.isfinite(sample["image"]).all()

    for key in ("pos", "cos", "sin", "width"):
        target = sample["targets"][key]
        assert target.shape == (1, 416, 416)
        assert torch.isfinite(target).all()

    print(f"{name}: {len(dataset)} samples — PASSED")

print("M6.2B DATASET SMOKE TEST PASSED")
```

# Utils

```python

from pathlib import Path
import shutil
import hashlib

RUN_NAME = "filmgrasp_colab_pilot01"

src = Path("/content/film-graspnet/runs") / RUN_NAME
dst = Path("/content/drive/MyDrive/0.InterviewPrep/FPT-Innovators/Assignment/FiLM-GraspNet/runs") / RUN_NAME

assert src.is_dir(), "Local training run not found"
assert not dst.exists(), "Drive run already exists; review before overwriting"

shutil.copytree(src, dst)

def sha256(path):
    digest = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()

required = [
    "checkpoints/last.pt",
    "checkpoints/best.pt",
    "metrics.csv",
    "config.json",
    "data_summary.json",
]

for name in required:
    a, b = src / name, dst / name
    assert a.is_file() and b.is_file(), f"Missing: {name}"
    assert sha256(a) == sha256(b), f"Checksum mismatch: {name}"
    print(f"{name}: VERIFIED")

print("GOOGLE DRIVE BACKUP PASSED")
```

```python

from pathlib import Path
import hashlib
import shutil

OLD_ROOT = Path("/content/drive/MyDrive/FiLM-GraspNet")
NEW_ROOT = Path(
    "/content/drive/MyDrive/0.InterviewPrep/"
    "FPT-Innovators/Assignment/FiLM-GraspNet"
)

RUN_NAME = "filmgrasp_colab_pilot01"

src = OLD_ROOT / "runs" / RUN_NAME
dst = NEW_ROOT / "runs" / RUN_NAME

# The previously verified backup should exist.
assert src.is_dir(), f"Source backup not found: {src}"
assert not dst.exists(), (
    f"Destination already exists. Review before copying: {dst}"
)

dst.parent.mkdir(parents=True, exist_ok=True)
shutil.copytree(src, dst)

def sha256(path):
    h = hashlib.sha256()
    with path.open("rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

source_files = {
    p.relative_to(src) for p in src.rglob("*") if p.is_file()
}
dest_files = {
    p.relative_to(dst) for p in dst.rglob("*") if p.is_file()
}

assert source_files == dest_files, "File inventory mismatch"

for relative_path in sorted(source_files):
    assert sha256(src / relative_path) == sha256(
        dst / relative_path
    ), f"Checksum mismatch: {relative_path}"

print("NEW DRIVE BACKUP VERIFIED")
print("Files verified:", len(source_files))
print("Location:", dst)
```
